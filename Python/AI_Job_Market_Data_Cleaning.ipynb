```python
df.columns
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(" ", "_")
)

df.columns
```

```python
text_columns = [
    "Job_Title",
    "Company",
    "Location",
    "Country",
    "Experience_Level",
    "Employment_Type",
    "Work_Mode",
    "Salary_Currency",
    "Industry",
    "Company_Size",
    "Education"
]

for col in text_columns:
    df[col] = df[col].astype("string").str.strip().str.title()
```

```python
df["Job_Title"].value_counts()
```

```python
df.isnull().sum()
```

```python
categorical_columns = [
    "Power_BI",
    "Tableau",
    "AWS",
    "Education",
    "Company_Size"
]

for col in categorical_columns:
    df[col] = df[col].fillna("unknown")
```

```python
df[["Salary_Min","Salary_Max","Years_Experience"]].dtypes
```

```python
numeric_columns = [
    "Salary_Min",
    "Salary_Max",
    "Years_Experience"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df["Posted_Date"] = pd.to_datetime(
    df["Posted_Date"],
    errors="coerce"
)
```

```python
df["Posted_Date"].dtype
```

```python
df["Average_Salary"] = (
    df["Salary_Min"] + df["Salary_Max"]
) / 2
```
